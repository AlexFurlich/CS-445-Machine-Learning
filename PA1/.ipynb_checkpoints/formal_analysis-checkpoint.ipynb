## Interpreting the plot

The plot demonstrates the range of values gathered through both training data and test data. It stops at 30 as that is when the training data hits 100% accuracy when given no limits on tree depth. 

**Underfitting** 
At the start of the graph we can see a significant issue where a small graph underfits the data. when only 5 or less levels of depth are allowed it is difficult for the decision tree to be able to properly adjust and provide enough specificity to correctly identify future test data. 

**Overfitting**
While the training data continues to improve as the levels of depth increase we can see that the test data reaches max accuracy at depth 10. After this the accuracy of the predictions diverge and we see the more layers of depth decrease the accuracy of the predicitons the decision tree provides. This is because we are too closely mapping to the exact training data provided. It becomes the perfect model for that training data specifically but much less useful for that general "catagory" of data.
