# `torch.utils.data.Dataset` and `torch.utils.data.DataLoader`

See also: https://pytorch.org/tutorials/beginner/basics/data_tutorial.html

In machine learning it is often the case that training data is too large to fit in memory on a single machine.  We may also want to perform some pre-processing on the data as it is loaded.  The `torch.utils.data.DataLoader` class provides a standard interface for feeding data to a machine learning model.  `DataLoader` objects are *iterables*: you loop over them with `for`, and you can loop over the same `DataLoader` again for every epoch.  (The `TensorDataset` used below keeps everything in memory.  For data that does not fit, you would write your own `Dataset` class that loads each element on demand.)

First let's generate some data using numpy:


```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from torch.utils.data import TensorDataset, DataLoader

np.random.seed(1)

def noisy_xor(num_points):
    """Synthetic 2d Dataset that is not linearly separable."""
    features = np.random.randint(2, size=(num_points, 2))
    labels = np.array(np.logical_xor(features[:, 0], features[:, 1]), dtype=np.float32)
    labels = np.expand_dims(labels, 1)
    features = np.array(
        (features + (np.random.random(features.shape))) / 2.0,
        dtype=np.float32,
    )
    return features, labels

features, labels = noisy_xor(20)
plt.plot(features[labels[:, 0] == 1, 0], features[labels[:, 0] == 1, 1], 's')
plt.plot(features[labels[:, 0] == 0, 0], features[labels[:, 0] == 0, 1], 'o')
plt.show()
```

Now we can create a `Dataset` object from our numpy arrays.  Each element of the dataset is a `(features, label)` pair:

```python
dataset = TensorDataset(torch.tensor(features), torch.tensor(labels))

print("\nIterate over the Dataset:\n")
for x, y in dataset:
    print(x, y)
```

## Batches

It is usually more efficient to process data in *batches* than individually. Here is an example of PyTorch code that multiplies each element in our data set by an appropriately sized weight vector and sums the result.  In this example each element is processed individually (**without** batches).

```python
total = torch.tensor([[0.0]])
weights = torch.tensor(np.random.random((2, 1)), dtype=torch.float32)

for x, y in dataset:
    total = total + x @ weights  # length 2 times 2x1, results in shape (1,)
    print(f"Total so far: {total}")

print(f"\nFinal Total: {total}")
```

Instead of processing one data element per iteration, we can batch the dataset and process multiple elements per iteration.  Many PyTorch operators, including the matrix multiplication operator, are "batch-aware" and will recognize that the first dimension corresponds to the batch.  Let's look at the first entry in a batched version of our dataset:

```python
loader_batched = DataLoader(dataset, batch_size=3, shuffle=True)
for X_batch, y_batch in loader_batched:
    print(f"Batch Shape: {X_batch.shape}\n")
    print("Features:\n", X_batch)
    print("Labels:\n", y_batch)
    break
```

Now we can re-calculate our weighted product using batches.  Because we passed `shuffle=True`, every loop over `loader_batched` visits the elements in a new random order.

Notice the last batch: 20 is not a multiple of 3, so it holds only 2 elements.  Code that processes batches should not assume every batch is full (or pass `drop_last=True` to skip the short one).

```python
total = torch.tensor([[0.0]])

for X_batch, y_batch in loader_batched:
    batch_of_products = X_batch @ weights  # 3x2 batch times 2x1 weight results in 3x1 result
    print(batch_of_products.shape)
    total = total + torch.sum(batch_of_products)
    print(f"Total so far: {total}")

print(f"\nFinal Total: {total}")
```

One thing to watch when you train with batches: if your loss *sums* over the elements of a batch, the size of each gradient step grows with the batch size, so the right learning rate depends on the batch size too.  Averaging instead (`torch.mean`, which is also the default in PyTorch's built-in loss functions) avoids that.
